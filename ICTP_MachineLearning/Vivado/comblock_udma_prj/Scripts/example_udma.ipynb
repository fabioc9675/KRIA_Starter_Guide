# Ejemplo UDMA

```python
 # MLab UDMA library
import udma

import numpy as np
from time import sleep
```

```python
IP_ADDRESS = '192.168.1.29'
IP_PORT = 7
```

```python
pynq_board = udma.UDMA_CLASS(IP_ADDRESS, IP_PORT)
```

```python
connectionStatus = 0
while(connectionStatus == 0):
    connectionStatus = pynq_board.connect()
    sleep(1)
```

```python
_ = pynq_board.log(0)
```

```python
REG_INPUT  = 2
REG_OUTPUT = 1
```

```python
data_in = pynq_board.read_reg(REG_OUTPUT)
print(data_in)
```

```python
data_in = pynq_board.read_reg(REG_INPUT)[1][0]
print(data_in)
```

```python
pynq_board.write_reg(REG_OUTPUT, 3)
```
